### **معیارهای ارزیابی مدل‌های طبقه‌بندی (Classification Metrics)**

در واقع، ماتریس درهم‌ریختگی (Confusion Matrix) اطلاعاتی درباره عملکرد مدل به ما می‌دهد و نشان می‌دهد که مدل چند نمونه را درست یا اشتباه طبقه‌بندی کرده است.

اما صرفاً شمردن تعداد پیش‌بینی‌های درست کافی نیست؛ چون ممکن است یک مدل در شرایط خاص عملکرد ظاهراً خوبی داشته باشد، درحالی‌که در تشخیص کلاس موردنظر چندان موفق نباشد.

برای همین، معیارهای دیگری مثل **Precision، Recall و F1-score** را معرفی می‌کنیم.

### **۱. Precision (دقت پیش‌بینی مثبت)**

این معیار به ما می‌گوید از میان تمام نمونه‌هایی که مدل آن‌ها را مثبت (Positive) تشخیص داده، چند مورد واقعاً مثبت بوده‌اند.

**فرمول:**

Precision = TP / (TP + FP)

- **TP (True Positive):** نمونه‌هایی که واقعاً مثبت بوده‌اند و مدل هم آن‌ها را مثبت تشخیص داده است.
- **FP (False Positive):** نمونه‌هایی که در واقع منفی بوده‌اند، اما مدل به اشتباه آن‌ها را مثبت تشخیص داده است.

به زبان ساده، Precision به این سؤال پاسخ می‌دهد:

**وقتی مدل می‌گوید یک نمونه مثبت است، چقدر می‌توانیم به پیش‌بینی آن اعتماد کنیم؟**

برای مثال، فرض کنید یک مدل تشخیص بیماری داریم. اگر مدل ۱۰۰ نفر را بیمار تشخیص دهد، اما فقط ۸۰ نفر واقعاً بیمار باشند، Precision برابر با ۸۰ درصد خواهد بود.

Precision = 80 / (80 + 20) = 0.8

**چه زمانی Precision اهمیت بیشتری دارد؟**

زمانی که مثبت کاذب (FP) هزینه زیادی داشته باشد. یعنی نمی‌خواهیم مدل نمونه‌ای را به اشتباه مثبت اعلام کند.

مثلاً:

- فیلتر تشخیص ایمیل اسپم: نمی‌خواهیم ایمیل مهمی را به اشتباه اسپم تشخیص دهیم.
- سیستم تشخیص تقلب بانکی: نمی‌خواهیم تراکنش‌های سالم را به اشتباه مسدود کنیم.
- سیستم تشخیص محتوای نامناسب: نمی‌خواهیم محتوای سالم را به اشتباه حذف کنیم.

البته یک نکته مهم وجود دارد: ممکن است مدلی تقریباً هیچ نمونه‌ای را مثبت تشخیص ندهد و در نتیجه Precision بسیار بالایی داشته باشد. حتی اگر فقط یک نمونه را مثبت تشخیص دهد و آن یک نمونه هم درست باشد، Precision آن ۱۰۰ درصد خواهد بود!

بنابراین Precision به‌تنهایی معیار کاملی برای ارزیابی مدل نیست؛ چون درباره تعداد نمونه‌های مثبتی که مدل از دست داده است، چیزی به ما نمی‌گوید.

### **۲. Recall (فراخوانی یا حساسیت)**

برای رفع این مشکل، معیار دیگری به نام Recall را معرفی می‌کنیم.

Recall به ما می‌گوید از میان تمام نمونه‌هایی که واقعاً مثبت بوده‌اند، مدل موفق شده چند مورد را به‌درستی شناسایی کند.

**فرمول:**

Recall = TP / (TP + FN)

- **TP (True Positive):** نمونه‌های مثبتی که مدل درست تشخیص داده است.
- **FN (False Negative):** نمونه‌های مثبتی که مدل به اشتباه منفی تشخیص داده است.

به زبان ساده، Recall به این سؤال پاسخ می‌دهد:

**از تمام نمونه‌های مثبتی که واقعاً وجود داشته‌اند، مدل چند مورد را پیدا کرده است؟**

مثلاً فرض کنید در یک مجموعه ۱۰۰ نفر واقعاً بیمار داریم. مدل از این تعداد فقط ۸۰ نفر را درست شناسایی می‌کند و ۲۰ نفر را سالم تشخیص می‌دهد.

در این صورت:

Recall = 80 / (80 + 20) = 0.8

**چه زمانی Recall اهمیت بیشتری دارد؟**

زمانی که منفی کاذب (FN) هزینه زیادی داشته باشد؛ یعنی از دست دادن یک نمونه مثبت برای ما خطرناک یا پرهزینه باشد.

مثلاً:

- تشخیص بیماری‌های خطرناک: نمی‌خواهیم فرد بیمار را سالم تشخیص دهیم.
- تشخیص تراکنش‌های مشکوک: نمی‌خواهیم تراکنش متقلبانه از فیلتر سیستم عبور کند.
- تشخیص ایمیل‌های اسپم: نمی‌خواهیم تعداد زیادی ایمیل اسپم را به‌عنوان ایمیل سالم شناسایی کنیم.

در چنین شرایطی، بهتر است Recall بالایی داشته باشیم، حتی اگر این موضوع باعث افزایش تعداد مثبت‌های کاذب شود.

### **۳. تفاوت Precision و Recall در یک نگاه**

در واقع، این دو معیار دو جنبه متفاوت از عملکرد مدل را بررسی می‌کنند:

| معیار     | سؤال اصلی                                  | خطای مهم |
| --------- | ------------------------------------------ | -------- |
| Precision | از پیش‌بینی‌های مثبت، چند مورد درست هستند؟ | FP       |
| Recall    | از مثبت‌های واقعی، چند مورد پیدا شده‌اند؟  | FN       |

برای درک بهتر، فرض کنید یک سیستم تشخیص بیماری داریم:

- اگر مدل فقط افرادی را بیمار اعلام کند که تقریباً مطمئن است بیمار هستند، Precision بالایی خواهد داشت.
- اگر مدل تلاش کند هیچ فرد بیماری را از دست ندهد، Recall بالایی خواهد داشت؛ حتی اگر تعدادی از افراد سالم را هم بیمار اعلام کند.

بنابراین معمولاً بین Precision و Recall یک بده‌بستان (Trade-off) وجود دارد. با تغییر آستانه تصمیم‌گیری مدل (Decision Threshold)، می‌توانیم تعادل متفاوتی بین این دو ایجاد کنیم.

### **۴. F1-score؛ ترکیب Precision و Recall**

در نهایت، برای داشتن یک معیار واحد که هر دو جنبه Precision و Recall را در نظر بگیرد، از معیار F1-score استفاده می‌کنیم.

**فرمول:**

F1 = 2 × (Precision × Recall) / (Precision + Recall)

این معیار در واقع میانگین هارمونیک (Harmonic Mean) دو معیار Precision و Recall است.

چرا میانگین هارمونیک؟

چون برخلاف میانگین حسابی، به مقادیر کوچک حساسیت بیشتری دارد. یعنی اگر یکی از دو معیار خیلی پایین باشد، F1-score هم کاهش قابل‌توجهی پیدا می‌کند.

برای مثال:

فرض کنید:

Precision = 0.9

Recall = 0.3

میانگین حسابی این دو برابر است با:

(0.9 + 0.3) / 2 = 0.6

اما F1-score برابر می‌شود با:

F1 = 2 × (0.9 × 0.3) / (0.9 + 0.3)

F1 = 0.54 / 1.2 = 0.45

همان‌طور که می‌بینیم، با وجود Precision بالا، Recall پایین باعث شده F1-score هم چندان بالا نباشد.

**چه زمانی از F1 استفاده کنیم؟**

زمانی که هر دو نوع خطای FP و FN برایمان اهمیت دارند و می‌خواهیم مدلی داشته باشیم که بین Precision و Recall تعادل برقرار کند.

این معیار به‌خصوص در مجموعه‌داده‌های نامتوازن (Imbalanced Datasets) کاربرد زیادی دارد؛ جایی که تعداد نمونه‌های یک کلاس بسیار بیشتر از کلاس دیگر است.

### **جمع‌بندی**

| معیار     | هدف                                | چه زمانی مهم‌تر است؟                                   |
| --------- | ---------------------------------- | ------------------------------------------------------ |
| Precision | کاهش مثبت‌های کاذب                 | وقتی اعلام اشتباه یک نمونه به‌عنوان مثبت هزینه‌بر است. |
| Recall    | کاهش منفی‌های کاذب                 | وقتی از دست دادن یک نمونه مثبت خطرناک است.             |
| F1-score  | ایجاد تعادل بین Precision و Recall | وقتی هر دو نوع خطا اهمیت دارند.                        |

**یک نکته مهم:** هیچ‌کدام از این معیارها به‌تنهایی بهترین معیار ارزیابی برای تمام مسائل نیستند. انتخاب معیار مناسب کاملاً به ماهیت مسئله و هزینه‌ای که هر نوع اشتباه برای ما ایجاد می‌کند بستگی دارد.

همچنین F1-score فقط Precision و Recall را در نظر می‌گیرد و True Negativeها را مستقیماً وارد محاسبه نمی‌کند؛ بنابراین برای ارزیابی کامل مدل، بهتر است در کنار Confusion Matrix و سایر معیارها بررسی شود.
