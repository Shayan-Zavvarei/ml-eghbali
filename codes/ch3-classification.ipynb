حالا کار سختش اینه که بتونیم به نحوی مغیار هایی رو درست کنیم که بتونه مدل رو ارزیابی کنه. یکی از راه هاش اینه که 

حالا سوال اینه که چحوری میشه یه کلسیفیکشن باینری رو ارزیابی کنیم؟ 
با 
CONFUSION MATRIX
هدف ما اینه ه بفهمیم یه سری پردیکشن ما ترو هست یا فالس 
TP FN
FP TN 

و ما یه سری مقادیر واقعی هم داریم که اینا هم همینن. یا پازتیو هستن یا نگتیو یعنی یه جا تصویر یا واقعا ۸ بوده و مدل پیشبینی کرده  که ۸ بوده
یه جا تصویر ۸ نبوده ولی مدل گفته ۸ بوده
یه جا تصویر ۸ بوده ولی مدل نگفته ۸ بوده
یه جا تصویر ۸ نبوده ولی مدل هم گفته ۸ نبوده
پس دو جا مدل درست پیشبینی کرده 
دو جا غلط
بر اساس این چهار تا عدد و تعداد تکرارشون میشه به نوعی مدل رو ارزیابی کرد و معیار بسازیم.

$$
Accuracy(ACC) = (TP + TN) / (TP + TN + FP + FN)
$$

که به این میگیم دقت. 
همین ماتریش رو میشه به تعداد بیشتر مثلا حتی هزار در هزار هم به دست آورد.

```python
# show the accuracy
sgd.score(X_train, y_train8)
```

```python
# using cross validation
from sklearn.model_selection import cross_val_score
cvs = cross_val_score(sgd, X_train, y_train8, cv =5, scoring="accuracy")
```

```python
cvs.mean()
```

که خب این شبیه هست به اون روش قبلیمون

```python
from sklearn.dummy import DummyClassifier 
dummy_c = DummyClassifier()
dummy_c.fit(X_train, y_train8)
dummy_c.predict(X_train)
```

این دامی صرفا میاد اون تعدادی که بیشتر هست توی دیتاستمون رو پیشبیینی میکنه که خب به درد نمیخوره
مثلا اینجا اونایی که هشت نیستن بیشتر از اونایی که هشت هستنه پس این دامی میاد اونا رو برمیگردونه

```python
dummy_c.predict(X_train).sum()
```

حالا بیایم این کراس ولیدیشن رو با این دامی کلسیفایر تست کنیم:

```python
cross_val_score(dummy_c,X_train,y_train8,cv = 5, scoring="accuracy").mean()
```
